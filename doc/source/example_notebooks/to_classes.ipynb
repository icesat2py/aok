## **Key ICESat-2 Variables/Datasets Used**

### **Geolocation Group** (`/{beam}/geolocation/`)
- `segment_id` - segment identification
- `reference_photon_lat` - latitude of reference photon
- `reference_photon_lon` - longitude of reference photon
- `ph_index_beg` - photon index beginning
- `segment_ph_cnt` - segment photon count
- `segment_dist_x` - along-track distance
- `segment_length` - segment length
- `delta_time` - transmit time
- `ref_elev` - reference elevation
- `ref_azimuth` - reference azimuth
- `solar_elevation` - solar elevation angle
- `solar_azimuth` - solar azimuth angle

### **Heights Group** (`/{beam}/heights/`)
- `h_ph` - photon heights
- `lat_ph` - photon latitude
- `lon_ph` - photon longitude
- `dist_ph_along` - along-track distance per photon
- `dist_ph_across` - across-track distance per photon
- `delta_time` - time per photon
- `signal_conf_ph` - signal confidence (photon level)
- `quality_ph` - photon quality flag

### **Background Atlas Group** (`/{beam}/bckgrd_atlas/`)
- `bckgrd_rate` - background photon rate
- `delta_time` - background measurement time

### **Geophysical Corrections Group** (`/{beam}/geophys_corr/`)
- `geoid` - geoid corrections

```python
def compare_dicts(dict1, dict2, name1="dict1", name2="dict2"):
    """
    Compare two dictionaries and print their keys, whether they are identical,
    and any differences in keys or values, including nested levels.
    
    Parameters:
    dict1 (dict): First dictionary to compare
    dict2 (dict): Second dictionary to compare
    name1 (str): Name for the first dictionary in output (default "dict1")
    name2 (str): Name for the second dictionary in output (default "dict2")
    """
    import numpy as np
    
    def objects_equal(obj1, obj2):
        if type(obj1) != type(obj2):
            return False
        if isinstance(obj1, dict):
            if set(obj1.keys()) != set(obj2.keys()):
                return False
            return all(objects_equal(obj1[k], obj2[k]) for k in obj1)
        elif isinstance(obj1, (list, tuple)):
            if len(obj1) != len(obj2):
                return False
            return all(objects_equal(a, b) for a, b in zip(obj1, obj2))
        else:
            try:
                return np.array_equal(obj1, obj2)
            except:
                return obj1 == obj2
    
    def print_diffs(d1, d2, path=""):
        for key in set(d1.keys()) | set(d2.keys()):
            full_path = f"{path}.{key}" if path else str(key)
            if key not in d1:
                print(f"Key {full_path} only in {name2}")
            elif key not in d2:
                print(f"Key {full_path} only in {name1}")
            elif not objects_equal(d1[key], d2[key]):
                if isinstance(d1[key], dict) and isinstance(d2[key], dict):
                    print_diffs(d1[key], d2[key], full_path)
                else:
                    print(f"Key {full_path} differs")
                    try:
                        print(f"  In {name1}: {d1[key]}")
                    except Exception as e:
                        print(f"  In {name1}: <error printing value: {e}>")
                    try:
                        print(f"  In {name2}: {d2[key]}")
                    except Exception as e:
                        print(f"  In {name2}: <error printing value: {e}>")
    
    print(f"Keys in {name1}:", sorted(dict1.keys()))
    print(f"Keys in {name2}:", sorted(dict2.keys()))
    print("Are keys equal?", set(dict1.keys()) == set(dict2.keys()))
    
    if objects_equal(dict1, dict2):
        print("Dictionaries are identical")
    else:
        print("Dictionaries differ")
        print_diffs(dict1, dict2)
```

```python
# Compare IS2_atl03_mds and IS2_atl03_mdsAI
compare_dicts(IS2_atl03_mds, IS2_atl03_mdsAI, "IS2_atl03_mds", "IS2_atl03_mdsAI")
```
